## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision it improves: Which pages a content/SEO team spends their limited review time on this
week, instead of guessing or working alphabetically.

Who acts on it: A content strategist or SEO lead with a fixed weekly review capacity — say, can
only manually check ~50 pages.

Action taken: Refresh stale copy, expand thin content, review CTR/title on visible-but-weak
pages, or monitor lower-priority pages.

Cost of a wrong recommendation: A false positive wastes review time on a page that wasn't
really declining — annoying but low-stakes. A false negative (missing a real decliner) means that
page keeps losing traffic unnoticed for longer, which is the more expensive failure since decay
compounds silently over time.

Unit of analysis: One row = one content page (content_id), measured over a 90-day window.

Why this isn't just "train a model": The model is one small piece. Most of the actual work is
in defining a fair decision (what does "worth reviewing" mean?), building a baseline a human could
already do by hand, checking the model doesn't just memorize noise or leak the answer, and turning
a probability into a ranked list a real reviewer can act on with limited time. A model with no
baseline, no validation, and no reason codes attached is not useful — it's a black box a reviewer
can't trust.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can claim: This is a decision-support tool. It ranks pages by evidence of decline,
staleness, or under-performance so a reviewer spends time efficiently. Precision@50 tells us how
many top-ranked pages are correctly flagged as declining under a current-window label — not proof
that fixing them causes recovery.

What I cannot claim: I cannot say a refresh will cause a page to recover — that needs a causal
experiment, not this data. I cannot claim to have reverse-engineered Google's ranking algorithm.
My current label (trend_direction == "down") is a same-window proxy, not a future outcome — a
stronger version of this lane, which I plan to build toward in later weeks, would predict future
decline from prior-only signals using the full warehouse data.
